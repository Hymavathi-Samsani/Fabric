```python
#Joins - Inner,left,right,outer,semileft,semiright,full

dept_path='abfss://hymaa_practice_2026@onelake.dfs.fabric.microsoft.com/lh_raw.Lakehouse/Files/dept.csv'
emp_path='abfss://hymaa_practice_2026@onelake.dfs.fabric.microsoft.com/lh_raw.Lakehouse/Files/emp.csv'

d_df=spark.read.csv(dept_path,header=True)
display(d_df)
e_df=spark.read.csv(emp_path,header=True)
display(e_df)
```

```python
#Inner Join - Returns rows with matching keys in both DataFrames.
display(d_df.join(e_df,d_df.deptid==e_df.deptno,'inner'))

#Full join - All rows from both sides. Non‑matches get NULL.
display(d_df.join(e_df,d_df.deptid==e_df.deptno,'full'))

#Left join - All rows from left DF, plus matches from right DF. Non‑matches get NULL
display(d_df.join(e_df,d_df.deptid==e_df.deptno,'left'))

#Right join - All rows from right DF, plus matches from left DF. Non‑matches get NULL.
display(d_df.join(e_df,d_df.deptid==e_df.deptno,'right'))

#Left-semi join - Returns only rows from left DF that have a match in right DF.
display(d_df.join(e_df,d_df.deptid==e_df.deptno,'left_semi'))

#Left-anti join - Returns only rows from left DF that do not have a match in right DF.
display(d_df.join(e_df,d_df.deptid==e_df.deptno,'left_anti'))

#Right-semi join - Returns only rows from left DF that have a match in right DF.
display(d_df.join(e_df,d_df.deptid==e_df.deptno,'semi'))

#Right-anti join - Returns only rows from left DF that have a match in right DF.
display(d_df.join(e_df,d_df.deptid==e_df.deptno,'anti'))

#Cross join - Cartesian product (every row of left with every row of right).
display(d_df.crossJoin(e_df))
```

```python
#Window Function - dense_rank(),rank(),row_number(),lead,lag

df=spark.read.csv('abfss://hymaa_practice_2026@onelake.dfs.fabric.microsoft.com/lh_raw.Lakehouse/Files/emp/emp_window_data.csv',header=True)
display(df)
df1=df
```

```python
from pyspark.sql.functions import *
from pyspark.sql.window import *

windowspec=Window.orderBy(df.sal.desc())
df=df.withColumn('dense_rank',dense_rank().over(windowspec))
df=df.withColumn('rank',rank().over(windowspec))
df=df.withColumn('row_number',row_number().over(windowspec))
display(df)

#Get 4th Highest salary records
df=df.filter(df.dense_rank==4)
display(df)
```

```python
display(df1)
```

```python
#Get dense_rank,rank and rows_number by department and get 3rd highest salary

df1=df1.withColumn('dense_rank',dense_rank().over(Window.partitionBy('deptno').orderBy(df1.sal.desc())))
df1=df1.withColumn('rank',rank().over(Window.partitionBy('deptno').orderBy(df1.sal.desc())))
df1=df1.withColumn('row_number',row_number().over(Window.partitionBy('deptno').orderBy(df1.sal.desc())))
display(df1)

#Get department-wise 3rd highest slaary records
df2=df1.filter(df1.dense_rank==3)
display(df2)
```

```python
#Get sales data
df=spark.read.csv(
    'abfss://hymaa_practice_2026@onelake.dfs.fabric.microsoft.com/lh_raw.Lakehouse/Files/Product-Sales-Region.csv',
    header=True)
display(df)
```

```python
#Apply transformations
df=df.withColumn('year',year(df.Date))
df=df.withColumn('quarter',quarter(df.Date))
display(df)
```

```python
df1=df.select('Region','Product','Quantity','UnitPrice','Year','quarter','StoreLocation')
display(df1)
```

```python
#create sales column
df1=df1.withColumn('sales',df1.UnitPrice*df1.Quantity)
display(df1)
```

```python
#Get year-wise-total quarter sales

df2=df1.groupBy(['Product','Region','Year']).agg(sum('sales').alias('Total Sales')).orderBy(*['Product','Region','year'])
display(df2)
```

```python
df2=df2.withColumn('Previous year total sales',lag('Total Sales',1,0).over(Window.partitionBy(['Product','Region']).orderBy(['year'])))
display(df2)
```

```python
df2=df2.withColumn('Sales Percentage Of Change',((col('Total Sales')-col('Previous year total sales'))/col('Previous year total sales'))*100)
display(df2)
```

```python
#Which year percetage of sales increased
df3=df2.filter(col('Sales Percentage Of Change') > 0).drop(col('Sales Percentage'))
display(df3)
```

```python
df4=df2.withColumn('Next year total sales',lead('Total Sales',1,0).over(Window.partitionBy(['Product','Region']).orderBy(['year']))).drop(col('Sales Percentage Of Change'))
# df2=df2.withColumn('Sales Percentage Of Change',((col('Total Sales')-col('Next year total sales'))/col('Next year total sales'))*100)

display(df4)
```

```python
#Collect_list and COllect_set
data=[
    ('A','apple'),
    ('A','banana'),
    ('B','orange'),
    ('B','grape'),
    ('C','kiwi')
]

df=spark.createDataFrame(data,['id','fruit'])
display(df)
```

```python
df1=df.groupBy('id').agg(collect_list('fruit').alias('fruits'))
display(df1)
```

```python
#Collect_list and COllect_set
data=[
    ('A','apple'),
    ('A','banana'),
    ('A','banana'),
    ('A','banana'),
    ('B','orange'),
    ('B','grape'),
    ('C','kiwi')
]

df=spark.createDataFrame(data,['id','fruit'])
display(df)

df1=df.groupBy('id').agg(collect_list('fruit').alias('fruits'))
display(df1)

df2=df.groupBy('id').agg(collect_set('fruit').alias('fruits'))
display(df2)
```
