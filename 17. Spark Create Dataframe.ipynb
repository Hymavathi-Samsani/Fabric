```python
# Ways to create dataframe

# 1. sqlContext --> sqlc.createDataframe
# 2. SparkSession  --> spark.createDataFrame
# 3. Spark read API --> spark.read
# 4. Apply transformations on dataframe --> result is dataframe
# 5. Convert RDD to dataframe
# 6. Using spark.sql
# 7. Convert pandas dataframe to spark dataframe
```

```python
# 1. sqlContext --> sqlc.createDataframe

from pyspark.sql import  SQLContext

lst=[('j','k'),('l','m'),('n','o')]
sqlc=SQLContext(sc)
df=sqlc.createDataFrame(lst)
df.show()
```

```python
# 2. SparkSession  --> spark.createDataFrame

from pyspark.sql import SparkSession

ss=SparkSession.builder.getOrCreate()

lst=[('j','k'),('l','m'),('n','o')]
df=ss.createDataFrame(lst)
df.show()
display(df)
```

```python
# Spark session

lst=[('j','k'),('l','m'),('n','o')]
df=spark.createDataFrame(lst)
df.show()
```

```python
# Create dataframe with row object

from pyspark.sql import Row

lst=[
    Row(id=1,name='Hymaa'),Row(id=2,name='Hemaa')
]

df=spark.createDataFrame(lst)
df.show()
```

```python
# Create dataframe using tuples

lst=[
    (101,'Ravi',22),
    (102,'Phani',23),
    (103,'Swetha',21)
]

df=spark.createDataFrame(lst,schema=['ID','Name','Age']) # Provide names to header using schema
df.show()
df.select('Name','Age').show() #Select only required columns
```

```python
#Convert dataframe to table in sql temporarely to access data through sq commands

df.createOrReplaceTempView('emp_v') # It is virtual view and physically not exists
```

```python
%%sql

select * from emp_v
```

```python
# 3. Spark read API --> spark.read

#type of files
#  csv
#  json
#  parquet
#  avaro
#  delta
# jdbc
# text

path='abfss://hymaa_practice_2026@onelake.dfs.fabric.microsoft.com/lh_raw.Lakehouse/Files/CSV/emp_details.csv'
df1=spark.read.format('csv').option('header',True).load(path)
display(df1)
```

```python
#Create temporary view for df1 dataframe to access through sql commands
df1.createOrReplaceTempView('emp_dtls_v')
```

```python
%%sql
---Access data through sql commands
select * from emp_dtls_v
```

```python
# 6. Using spark.sql
#Access temporary view data with sql commands and return output format is dataframe

df2=spark.sql('select * from emp_dtls_v')
display(df2)
```

```python
# 4. Apply transformations on dataframe --> result is dataframe

#withColumn method - to create new column

from pyspark.sql.functions import lit

#Lit is a function used fill entire column cels with constant value
df2=df1.withColumn('Country',lit('IND'))
display(df2)
type(df2)
```

```python
# 5. Convert RDD to dataframe

#1st method
lst=[(1,'ravi'),(2,'phani')]
rdd=sc.parallelize(lst)

df=spark.createDataFrame(rdd,['Id','Name'])
display(df)

#2nd method
from pyspark.sql import Row
rdd1=rdd.map(lambda item: Row(id=item[0],name=item[1]))
df1=rdd1.toDF()
display(df1)
```

```python
# 7. Convert pandas dataframe to spark dataframe and viceversa

import pandas as pd

lst=[(1,'ravi'),(2,'phani')]

df=pd.DataFrame(lst)
print(df,'Type is - ',type(df))
```

```python
df1=spark.createDataFrame(df)
print(display(df1),'Type is - ',type(df1))
```

```python
#Convert spark dataframe to pandas dataframe
df2=df1.toPandas()
print(df2,'Type is - ',type(df2))
```
