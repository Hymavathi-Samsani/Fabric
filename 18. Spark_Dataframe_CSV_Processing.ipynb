```python
#Spark Dataframe CSV Processing
# #Options - header
#          - inferschema
#          - sep
#          - mode
#          - badrecordspath
#          - custom schema
#          - recursiveFileLookup

path='abfss://hymaa_practice_2026@onelake.dfs.fabric.microsoft.com/lh_raw.Lakehouse/Files/CSV/emp_details.csv'
df=spark.read.format('csv').option('header',True).load(path)
display(df)
print('\nDataFrame columns are - ',df.columns)
```

```python
#Inferschema - change column data type based on data values it is holding
df=spark.read.format('csv').option('header',True).option('inferschema',True).load(path)
display(df)
print(df.columns)
```

```python
#Other way to read csv data

df=spark.read.csv(path,header=True,inferSchema=True)
display(df)
```

```python
# When we have millions of records and using inferschema scanning all records
# to decide the data type is very time consuming processing. Hence, we can create custom schema for the dataframe using below sytax

#Creating custom schema
#['id', 'name', 'salary', 'gender', 'deptno']

from pyspark.sql.types import *

cust_schema=StructType(
    [
         StructField('id',IntegerType()),
         StructField('name',StringType()),
         StructField('salary',IntegerType()),
         StructField('gender',StringType()),
         StructField('deptno',StringType())
    ]
)

#Apply Custom Schema

df=spark.read.format('csv').option('header',True).schema(cust_schema).load(path)
display(df)
```

```python
#How to hande bad records in dataframe - Applying custom schema

path='abfss://hymaa_practice_2026@onelake.dfs.fabric.microsoft.com/lh_raw.Lakehouse/Files/CSV/emp_badrecords.csv'
df=spark.read.format('csv').option('header',True).load(path)
display(df)
```

```python
#Here you can observe that bad records have been set as null after applying custom schema
from pyspark.sql.types import *

cust_schema=StructType(
    [
         StructField('id',IntegerType()),
         StructField('name',StringType()),
         StructField('gender',StringType()),
         StructField('salary',IntegerType())
    ]
)

df=spark.read.format('csv').option('header',True).schema(cust_schema).load(path)
display(df)
```

```python
#Moving al bad records to particular path
path='abfss://hymaa_practice_2026@onelake.dfs.fabric.microsoft.com/lh_raw.Lakehouse/Files/CSV/emp_badrecords.csv'

df=spark.read.format('csv')\
   .option('header',True)\
   .schema(cust_schema)\
   .option('badrecordspath','abfss://hymaa_practice_2026@onelake.dfs.fabric.microsoft.com/lh_raw.Lakehouse/Files/CSV/tmp_badrecords')\
   .load(path)

display(df)

#OPtion - badrecordspath - exclusive for databricks and won't work in fabric - similar one have to find for fabric
```

```python
#Modes to handle bad records in fabric
# There are 3 types of modes
#  1. PERMISSIVE - It is default mode. It shows null values
#  2. DROPMALFORMED - Drops the records which have bad data
#  3. FAILFAST - Fail the spark dataframe

path='abfss://hymaa_practice_2026@onelake.dfs.fabric.microsoft.com/lh_raw.Lakehouse/Files/CSV/emp_badrecords.csv'

#PERMISSIVE
df=spark.read.format('csv')\
    .option('header',True)\
    .schema(cust_schema)\
    .option('mode','PERMISSIVE')\
    .load(path)

print('PERMISSIVE MODE - Replaces bad data with NULL')
df.show()

#DROPMALFORMED
df1=spark.read.format('csv')\
    .option('header',True)\
    .schema(cust_schema)\
    .option('mode','DROPMALFORMED')\
    .load(path)

print('DROPMALFORMED MODE - Drops all records which have bad data')
df1.show()

#FAILFAST
df2=spark.read.format('csv')\
    .option('header',True)\
    .schema(cust_schema)\
    .option('mode','FAILFAST')\
    .load(path)

print('FAILFAST MODE - Show error or fail spark dataframe')
df2.show()
```
