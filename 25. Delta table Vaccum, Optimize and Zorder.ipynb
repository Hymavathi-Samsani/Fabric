```python

# Delta Tables - Support Updates
#              - Versioning
#                      - Timetravel
#                      - Schema evolution - Support dynamic schema changes


# Tables are nothing but files at the backend.. so, in versioning lot of files be created and each time system go through every file to get active records
# hence, performance gets affected.

# To manage this we can use 

#          - Optimize - It merges all active record files into single file. So, system will scan only this one file give one file and give results
#          - Vaccum - It removes unused part files based hours created at the backend with delta versioning. You can mention how manys days files have to remove.
#            Also, you can check which files going to remove in vaccum using dry run. 
#          - Zorder - When you multiple files this command sorts all files data sequentially based on highest cordinality column and creates new files
#            and store each file information with min and max one main table. So when query data it will scan the main table first and get the information about
#            which file(s) to scan to give the results back



# ****************CLONING DELTA TABLE VERSIONING NOTEBOOK **************************
```
