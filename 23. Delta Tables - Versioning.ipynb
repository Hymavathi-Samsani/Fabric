```python
# Database vs Datawarehouse vs datalake vs datalakehouse

# Database - Think like - Operation system
#          - Purpose - Store and manage structured data for day to day operations (OLTP) 
#          - Datatype - Highly Structured data  like tables, columns,rows
#          - Usecase - Apllic ation like e-commerce,banking and HR systems
#          - Example - SQL server, myqsl,postgreSQL

# Datawarehouse - Think like - Analytic System
#               - Purpose - Central repository for structured data, historical data optimized for analytics
#               - Datatype - Structured,cleaned and transformed data 
#               - Usecase - Bussiness Intelligence, dashboards and reporting
#               - Example - Azure Synapse, Snowflake, Amazon Redshift

# Datalake - Think like - Raw storage
#          - Purpose - Store raw and large scale data in its native formate(structured, unstructured,Semi-structured)
#          - Datatype - CSV, JSOn,Parquet,video,images,logs
#          - Usecase - Big Data storage, machine learning and exploratory analytics
#          - Example - Azure datalake,Amazon S3

# Datalakehouse - Think like - The modren unified platforma that merges all the three worlds
#          - Purpose - Hybrid approach combining data lake flexibility + datawarehouse performance
#          - Datatype - Supports Structured, Semi-Structured, Unstructured
#          - Usecase - Unified platform for BI+ML+Advance analytics
#          - Example - Microsoft Fabric lakehouse, Databricks delta lake


# Delta Tables - Support Updates
#              - Versioning
#                      - Timetravel
#                      - Schema evolution - Support dynamic schema changes

```
